**Movilidad observada ≠ demanda de autobús ≠ rentabilidad comercial.**

La existencia de una ruta privada permite estudiar sus características, pero **no** demuestra por sí sola su rentabilidad. Toda conclusión debe indicar qué datos la respaldan y qué incertidumbre conserva.

# 1. Visión del proyecto

**Estado:** En elaboración · **Documento origen:** `00_ideas_busmatch.md` (propuesta inicial del producto).

### 1.1. Problema que abordamos

### 1.1. Definición del problema

Las empresas de transporte de viajeros se enfrentan a un escenario de incertidumbre a la hora de identificar nuevos corredores en los que podría resultar interesante establecer servicios regulares de transporte colectivo.

Tradicionalmente, la identificación de estas oportunidades puede depender de solicitudes comerciales de empresas, instituciones u otros clientes que manifiestan una necesidad concreta de transporte. Sin embargo, cuando estas solicitudes no existen, los operadores pueden carecer de información suficiente para detectar de manera proactiva territorios con patrones de movilidad que justifiquen estudiar la implantación de nuevos servicios.

**BusMatch pretende abordar esta problemática mediante el análisis sistemático de la movilidad de la población**, estudiando distintos corredores dentro de un ámbito geográfico determinado. A partir de fuentes de datos de movilidad y una metodología reproducible, el proyecto buscará caracterizar el volumen de desplazamientos, su recurrencia, su distribución temporal y otros indicadores relevantes.

El propósito es **reducir la incertidumbre en la fase inicial de exploración de oportunidades**, proporcionando información objetiva que permita identificar y priorizar corredores susceptibles de un estudio más profundo.

No obstante, es fundamental establecer una distinción entre **movilidad observada y demanda comercial potencial**. Aunque el estudio se desarrolle con rigor metodológico, los desplazamientos identificados no implican automáticamente que exista disposición a utilizar o contratar un servicio de autobús, ni permiten garantizar su viabilidad económica.

Por tanto, BusMatch no pretende asegurar la rentabilidad de los corredores identificados, sino ofrecer una base analítica sólida que facilite la toma de decisiones y oriente las posteriores fases de evaluación operativa y comercial.

### 1.2. Usuarios objetivo

- **Principal:** empresas de transporte de viajeros que buscan oportunidades de servicio.
- **Potenciales usuarios secundarios:** centros de actividad, empresas o agrupaciones interesadas en estudiar necesidades de transporte.

*Pendiente de concretar:* perfil exacto del usuario inicial (operador local, gestor de movilidad, empresa industrial, etc.) y la decisión específica que tomaría con los resultados.

### 1.3. Propuesta de valor (borrador)

Convertir fuentes agregadas de movilidad y datos geográficos en **una lista explicable y priorizada de corredores candidatos a investigación**, reduciendo el trabajo exploratorio previo a un estudio comercial.

### 1.4. Hipótesis de investigación

- **H1 (principal, por contrastar):** los corredores donde operan rutas privadas documentadas presentan patrones geográficos y temporales que pueden caracterizarse mediante datos públicos de movilidad.
- **H2 (por contrastar):** algunos corredores todavía no documentados como rutas privadas muestran patrones comparables a los de los corredores de referencia.
- **H3 (posterior):** cruzar esos candidatos con la oferta disponible permite identificar posibles necesidades de transporte insuficientemente atendidas.

**No es una hipótesis demostrada:** que una puntuación alta del modelo implique rentabilidad, ocupación o intención de compra.


# 2. Objetivos y alcance

**Estado:** En elaboración.

### 2.1. Objetivo general (propuesta)

Analizar datos públicos de movilidad para caracterizar corredores territoriales, identificar patrones asociados a servicios privados de autobús documentados y generar candidatos priorizados para su evaluación posterior.

### 2.2. Objetivos específicos (borrador)

1. Construir una base **Gold** de indicadores de movilidad, geográfica y temporalmente coherente.
2. Definir un inventario verificable de rutas privadas y sus características observables.
3. Estudiar si los patrones de movilidad de esas rutas permiten definir un método de priorización de corredores.
4. Evaluar el método frente a alternativas sencillas e identificar sus incertidumbres.
5. Contrastar la cobertura de transporte existente en los corredores priorizados, en la medida en que haya datos disponibles.
6. Documentar los requisitos de un estudio económico preliminar para fases posteriores.

### 2.3. Alcance inicial

| Dimensión | Propuesta inicial | Por decidir / validar |
|---|---|---|
| Territorio | Comarca de Torrijos y entorno de Toledo | Escala geográfica exacta y ampliación para obtener suficientes rutas de referencia |
| Periodo | Varias fechas comparables de movilidad | Fechas definitivas, cobertura y consistencia metodológica |
| Unidad de análisis | Corredor origen–destino, sentido y franja temporal | Tratamiento de recorridos con paradas intermedias |
| Mercado objetivo | Transporte privado de viajeros por autobús | Separar rutas corporativas de venta abierta y otros modelos |
| Entregable principal del TFM | Gold + método evaluado de selección de candidatos | Alcance real de fases 3 y 4 |

### 2.4. Fuera de alcance, salvo ampliación acordada

- Afirmar **rentabilidad demostrada** sin costes e ingresos observados.
- Predecir viajeros de un servicio nuevo sin validación de demanda.
- Gestionar reservas, cobros, flotas o servicios en operación.
- Atribuir viajes agregados a personas o empresas concretas.
- Presentar rutas concesionadas y rutas privadas como si fueran el mismo modelo económico.

- **Fuentes y linaje:** archivo original, versión metodológica, transformación y fecha.
- **Calidad:** nulos, privacidad/supresión, coberturas, agregación y consistencia temporal.

### Fundamento metodológico de la fase 2

Ante la falta de datos económicos de los operadores, utilizaremos la **permanencia de rutas privadas no subvencionadas como indicador indirecto de viabilidad comercial**.

Construiremos una base histórica de estas rutas y sus paradas, identificaremos los corredores que conectan y cruzaremos sus características con los datos de movilidad y del territorio.

Mediante técnicas estadísticas o de aprendizaje automático, buscaremos **patrones comunes que permitan identificar y clasificar nuevos corredores con características similares**.

El modelo deberá ofrecer resultados explicables, indicando qué factores favorecen cada candidatura y qué incertidumbres existen.

### Consideraciones metodológicas

- **Datos positivos:** las rutas consolidadas aportan evidencia de continuidad, pero la ausencia de una ruta no demuestra inviabilidad.
- **Aprendizaje:** se estudiarán técnicas adecuadas para trabajar sin ejemplos negativos fiables, como *Positive-Unlabeled Learning*.
- **Validación:** se tendrán en cuenta la cobertura de los datos, los sesgos de selección y, cuando sea posible, rutas históricas desaparecidas.
- **Interpretación:** el modelo identifica patrones asociados a la continuidad comercial, no demuestra rentabilidad ni causalidad.

### Dependencia entre fases

La recopilación histórica de rutas privadas comenzará **en paralelo con la construcción de la base Gold de la fase 1**, para disponer de los datos necesarios al iniciar la fase 2.

**Regla de paso:** cada fase entregará resultados e incertidumbres explícitos. La similitud con corredores consolidados permite priorizar candidatos, pero **no garantiza su viabilidad económica**, que se estudiará en la fase 4.

# 5. Metodología por fase

**Estado:** Diseño preliminar. Las técnicas se concretarán según los datos disponibles.

## 5.1. Fase 1 — Estudio de corredores

**Objetivo:** Comprender cómo se desplaza la población e identificar patrones de movilidad.

**Metodología:** Recopilar, limpiar y unificar los datos de movilidad del Ministerio para construir una base Gold y analizar los desplazamientos entre territorios.

**Resultado:** Base de datos validada y caracterización de los corredores.

**Limitaciones:** Diferencias temporales y geográficas, calidad y granularidad de los datos.

## 5.2. Fase 2 — Detección de candidatos

**Objetivo:** Identificar qué características comparten los corredores con rutas privadas consolidadas y localizar nuevos corredores con patrones similares.

**Metodología:**
1. Construir un inventario histórico de rutas privadas no subvencionadas, sus paradas y su continuidad.
2. Identificar los corredores asociados y cruzarlos con los datos de movilidad y características territoriales.
3. Aplicar análisis estadístico o técnicas de ML para descubrir patrones comunes y evaluar nuevos corredores.

**Resultado:** Ranking explicable de corredores candidatos, indicando los factores que justifican su selección y el grado de incertidumbre.

**Limitaciones:** La permanencia de una ruta es un indicador indirecto de viabilidad, no una prueba de rentabilidad. La ausencia de rutas no demuestra inviabilidad.

**Alternativa:** Si los datos son insuficientes, utilizar reglas de selección interpretables o técnicas de similitud, sin forzar modelos predictivos.
